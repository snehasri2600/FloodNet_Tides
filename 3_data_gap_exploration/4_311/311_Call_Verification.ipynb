```python
flooding_311['created_date'] = pd.to_datetime(flooding_311['created_date'])
flooding_311['date'] = flooding_311['created_date'].dt.date
```

```python
sensors = pd.read_csv('../../final_deployed_sensors.csv')[['sensor_name', 'latitude', 'longitude']]

qc = pd.read_csv('../FloodNet_Data_Manual_QC.csv')
qc['date'] = pd.to_datetime(qc['Date']).dt.date

qc_long = qc.assign(Sensors=qc['Sensors'].str.split(',\s*')).explode('Sensors')
qc_long['sensor_name'] = qc_long['Sensors'].str.strip()
qc_long = qc_long.merge(sensors, on='sensor_name', how='left')
```

```python
reports = flooding_311[['date', 'latitude', 'longitude', 'incident_address', 'descriptor']].dropna(subset=['latitude', 'longitude']).copy()
reports['latitude'] = reports['latitude'].astype(float)
reports['longitude'] = reports['longitude'].astype(float)
```

```python
import numpy as np

def haversine_miles(lat1, lon1, lat2, lon2):
    lat1, lon1, lat2, lon2 = map(np.radians, [lat1, lon1, lat2, lon2])
    dlat, dlon = lat2 - lat1, lon2 - lon1
    a = np.sin(dlat / 2)**2 + np.cos(lat1) * np.cos(lat2) * np.sin(dlon / 2)**2
    return 2 * 3958.8 * np.arcsin(np.sqrt(a))
```

```python
RADIUS_MILES = 0.5

def nearest_report(row):
    if pd.isna(row['latitude']):
        return pd.Series({'nearest_311_miles': np.nan, 'nearest_311_address': None, 'nearest_311_descriptor': None})
    same_day = reports[reports['date'] == row['date']]
    if same_day.empty:
        return pd.Series({'nearest_311_miles': np.nan, 'nearest_311_address': None, 'nearest_311_descriptor': None})
    dists = haversine_miles(row['latitude'], row['longitude'], same_day['latitude'], same_day['longitude'])
    idx = dists.idxmin()
    return pd.Series({
        'nearest_311_miles': dists.loc[idx],
        'nearest_311_address': same_day.loc[idx, 'incident_address'],
        'nearest_311_descriptor': same_day.loc[idx, 'descriptor'],
    })

qc_long[['nearest_311_miles', 'nearest_311_address', 'nearest_311_descriptor']] = qc_long.apply(nearest_report, axis=1)
qc_long['corroborated'] = qc_long['nearest_311_miles'] <= RADIUS_MILES

def nearby_reports(row):
    if pd.isna(row['latitude']):
        return []
    same_day = reports[reports['date'] == row['date']]
    if same_day.empty:
        return []
    dists = haversine_miles(row['latitude'], row['longitude'], same_day['latitude'], same_day['longitude'])
    return same_day.loc[dists <= RADIUS_MILES, 'incident_address'].tolist()

qc_long['nearby_311_addresses'] = qc_long.apply(nearby_reports, axis=1)
qc_long['num_nearby_311'] = qc_long['nearby_311_addresses'].str.len()

display(qc_long.head())
print(qc_long['corroborated'].value_counts())
```

## Bringing in rain and tide/surge corroboration

311 alone is a weak signal (biased toward denser, higher-reporting areas — see above).
`Tidal_Corroboration.ipynb` already built a `rain_flag` (rain at the sensor's nearest
weather station, day-of or day-before) and `tide_class` (whether that day's high tide
was unusually high astronomically, surged, both, or neither) for every QC event. Merge
that in here instead of recomputing it, so each bump can be judged on rain, tide/surge,
and 311 together.

One row — `SI - Grimsby St/Mapleton Ave` on 2025-12-19 — is missing from that file: it
was already flagged there as a QC slip (actually in the public feed on time, not a real
gap), so its tidal columns will come back NaN. That's expected, not a merge bug.

```python
tidal_events = pd.read_csv('../3_tidal/tidal_corroboration_events.csv', parse_dates=['date'])
tidal_events['date'] = tidal_events['date'].dt.date

qc_long = qc_long.merge(
    tidal_events[['date', 'sensor_name', 'rain_flag', 'tide_class', 'astro_pctile', 'surge_ft_h']],
    on=['date', 'sensor_name'], how='left', indicator='_tidal_merge'
)

qc_long['known_qc_slip'] = (
    (qc_long['sensor_name'] == 'SI - Grimsby St/ Mapleton Ave')
    & (qc_long['date'] == pd.Timestamp('2025-12-19').date())
)

# anything unmatched that ISN'T the known slip would be a real problem worth investigating
unexpected_unmatched = qc_long[(qc_long['_tidal_merge'] == 'left_only') & ~qc_long['known_qc_slip']]
print(f"unexpected unmatched rows: {len(unexpected_unmatched)}")
qc_long = qc_long.drop(columns='_tidal_merge')
```

## Three signals, side by side

Rain and tide/surge don't apply to the same sensors — a pluvial (street/sewer) sensor
isn't expected to show a tide signal, and a coastal sensor isn't expected to show rain.
So instead of averaging the three checks into one score, keep them as separate columns
and surface only the bumps where **none** of the three signals fired — those are the
ones with no independent corroboration at all, and worth a second look (or acceptance
that FloodNet caught something none of the other signals were positioned to see).

```python
qc_long['tide_signal'] = qc_long['tide_class'].notna() & (qc_long['tide_class'] != 'neither')
qc_long['num_signals'] = (
    qc_long['rain_flag'].fillna(False).astype(bool).astype(int)
    + qc_long['tide_signal'].astype(int)
    + qc_long['corroborated'].fillna(False).astype(bool).astype(int)
)

summary_cols = ['Date', 'sensor_name', 'rain_flag', 'tide_class', 'num_nearby_311', 'corroborated',
                 'nearest_311_miles', 'nearest_311_descriptor',
                 'num_signals', 'known_qc_slip']
display(qc_long[summary_cols].sort_values('num_signals'))

print(qc_long['num_signals'].value_counts().sort_index())
print()
no_signal = qc_long[(qc_long['num_signals'] == 0) & ~qc_long['known_qc_slip']]
print(f"{len(no_signal)} bumps have zero corroboration across rain, tide/surge, and nearby 311:")
display(no_signal[summary_cols])
```

```python
two_of_three = qc_long[qc_long['num_signals'] == 2]
print(f"{len(two_of_three)} bumps have 2 of 3 signals corroborating:")
display(two_of_three[summary_cols])
```

## Mapping corroboration: where and when

`num_signals` (0-3) is a magnitude, not a category, so both views below use one
sequential blue ramp (light = fewer signals, dark = more) instead of arbitrary colors.

- **Static map** — one marker per sensor (11 total), colored by that sensor's *mean*
  `num_signals` across all its bumps, sized by how many bumps it had. Shows which
  sensors tend to be well- vs. poorly-corroborated overall.
- **Slider map** — every individual bump plotted at its sensor's location, with a
  time slider (via `folium.plugins.TimestampedGeoJson`) stepping through 2025 day by
  day. Points accumulate as the slider advances, so dragging it shows *when* each
  sensor's bumps happened and whether corroboration clusters in time (e.g. a wet
  season) or is scattered — the same "where vs. when" split as a static map + chart,
  but keeping both dimensions inside folium since that's the renderer that's
  actually working in this notebook.

```python
sensor_summary = qc_long.groupby('sensor_name').agg(
    n_bumps=('num_signals', 'size'),
    mean_signals=('num_signals', 'mean'),
    latitude=('latitude', 'first'),
    longitude=('longitude', 'first'),
).reset_index()

display(sensor_summary.sort_values('mean_signals'))
```

```python
import folium
import branca.colormap as cm

colormap = cm.linear.Blues_09.scale(0, 3)
colormap.caption = 'Mean # corroborating signals (0-3)'

m = folium.Map(
    location=[sensor_summary['latitude'].mean(), sensor_summary['longitude'].mean()],
    zoom_start=10, tiles='cartodbpositron'
)

for _, row in sensor_summary.iterrows():
    folium.CircleMarker(
        location=[row['latitude'], row['longitude']],
        radius=6 + row['n_bumps'] / 4,
        color=colormap(row['mean_signals']),
        fill=True,
        fill_color=colormap(row['mean_signals']),
        fill_opacity=0.85,
        weight=1,
        tooltip=f"{row['sensor_name']} | {row['n_bumps']} bumps | mean signals: {row['mean_signals']:.2f}"
    ).add_to(m)

colormap.add_to(m)
m
```

```python
from folium.plugins import TimestampedGeoJson

def signals_fired(row):
    fired = []
    if bool(row['rain_flag']) if pd.notna(row['rain_flag']) else False:
        fired.append('rain')
    if row['tide_signal']:
        fired.append(f"tide/surge ({row['tide_class']})")
    if row['corroborated']:
        fired.append(f"311 nearby ({row['num_nearby_311']})")
    return ', '.join(fired) if fired else 'none'

def to_feature(row):
    color = colormap(row['num_signals'])
    return {
        'type': 'Feature',
        'geometry': {'type': 'Point', 'coordinates': [row['longitude'], row['latitude']]},
        'properties': {
            'time': row['date'].isoformat(),
            'popup': (
                f"{row['sensor_name']}<br>{row['Date']}<br>"
                f"signals fired ({row['num_signals']}/3): {signals_fired(row)}"
            ),
            'icon': 'circle',
            'iconstyle': {'fillColor': color, 'color': color, 'fillOpacity': 0.8, 'radius': 6},
        },
    }

bump_points = qc_long.dropna(subset=['latitude', 'longitude'])
features = bump_points.apply(to_feature, axis=1).tolist()

m2 = folium.Map(
    location=[bump_points['latitude'].mean(), bump_points['longitude'].mean()],
    zoom_start=10, tiles='cartodbpositron'
)

TimestampedGeoJson(
    {'type': 'FeatureCollection', 'features': features},
    period='P1D',
    add_last_point=True,
    auto_play=False,
).add_to(m2)

colormap.add_to(m2)
m2
```

## Conclusion

100/227 (44%) of the manually-flagged bumps have at least one independent corroborating
signal — rain, an unusually high/surging tide, or a nearby same-day 311 call — and
31/227 have two or three. That's evidence these bumps aren't noise: they're real
weather/tide events lining up with what FloodNet logged.

The 127 with zero signals (126 once the one known QC slip is set aside) aren't
evidence of the opposite. Rain doesn't apply to a coastal sensor and tide/surge
doesn't apply to a pluvial one, so a sensor only ever has access to one real physical
signal plus a weak, reporting-biased 311 check — three checks were never going to
cover every sensor. Zero signals most often means "these particular checks weren't
positioned to see it," not "this didn't happen."

Combined with the independent feed-verification pass in `Weather_Data_Scrape.ipynb`
(62 of 63 rain-driven QC events confirmed genuinely absent from the public feed, only
1 flagged as a QC slip and not a real gap), I'm treating the 2025 data gap as real:
these are actual floods missing from FloodNet's public dataset, not manual-QC
false positives.
